```sql
%python
print("Hello Hrashit")
print("GM all")

"""
Python All-in-One Quick Reference Script
Covers: Built-in Functions, String/List/Dict/Set Methods, Lambdas, File Ops, Exceptions
"""

# ==========================================
# 1. BUILT-IN DATA TYPES & TYPE CASTING
# ==========================================
num_int = 10
num_float = 20.5
num_str = "123"

# type(), int(), float(), str(), bool()
converted_int = int(num_str)
converted_float = float(num_int)
converted_str = str(num_float)
boolean_val = bool(1)  # 0 -> False, baki sab True

print("Type check:", type(converted_int), type(boolean_val))


# ==========================================
# 2. NUMERIC FUNCTIONS & MATH
# ==========================================
numbers = [-15, 3, 42, -8, 100, 25]

print("Length (len):", len(numbers))
print("Maximum (max):", max(numbers))
print("Minimum (min):", min(numbers))
print("Total Sum (sum):", sum(numbers))
print("Absolute Value (abs):", abs(-45))
print("Round (round):", round(3.14159, 2))
print("Power (pow):", pow(2, 4))  # 2^4 = 16
quotient, remainder = divmod(17, 5)  # (3, 2)
print("Divmod (17 // 5, 17 % 5):", quotient, remainder)


# ==========================================
# 3. STRING METHODS
# ==========================================
text = "  hello data engineers, welcome to python!  "

cleaned = text.strip()               # spaces hatata hai
upper_text = cleaned.upper()         # HELLO...
lower_text = cleaned.lower()         # hello...
title_text = cleaned.title()         # Capitalize Each Word
replaced = cleaned.replace("python", "Databricks")
words = cleaned.split(" ")           # List me split karega
joined = "-".join(["a", "b", "c"])   # "a-b-c"
starts = cleaned.startswith("hello") # True / False
ends = cleaned.endswith("!")        # True / False
found_index = cleaned.find("data")   # Index dhundega (warna -1)
count_e = cleaned.count("e")         # Count karega kitni baar aaya

print("Cleaned String:", cleaned)
print("Joined String:", joined)


# ==========================================
# 4. LIST & TUPLE METHODS
# ==========================================
my_list = [5, 2, 8, 1, 9]

my_list.append(10)          # Akhir me add karega
my_list.insert(1, 99)       # Index 1 par 99 insert karega
my_list.extend([11, 12])    # Multiple items add karega
popped_val = my_list.pop()  # Last element nikalega aur return karega
my_list.remove(99)          # Specified value ko delete karega

# Sorting & Reversing
sorted_copy = sorted(my_list) # Nayi sorted list deta hai (original safe)
my_list.sort(reverse=True)    # In-place descending sort karta hai
reversed_list = list(reversed(my_list))

print("Sorted List:", sorted_copy)


# ==========================================
# 5. DICTIONARY & SET METHODS
# ==========================================
# Dictionary
emp_data = {"id": 101, "name": "Harshit", "role": "Engineer"}

keys = list(emp_data.keys())       # saari keys
values = list(emp_data.values())   # saari values
items = list(emp_data.items())     # key-value pairs (tuples)
role = emp_data.get("role", "NA")  # safe access (KeyError nahi aayega)
emp_data.update({"city": "Udaipur"})

# Set (Unique values & Set operations)
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
union_set = set_a.union(set_b)             # {1, 2, 3, 4, 5, 6}
intersect_set = set_a.intersection(set_b)  # {3, 4}
diff_set = set_a.difference(set_b)         # {1, 2}


# ==========================================
# 6. ITERATION HELPERS: enumerate(), zip(), range()
# ==========================================
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

# enumerate(): index aur value dono sath me deta hai
for idx, name in enumerate(names, start=1):
    print(f"Index {idx}: {name}")

# zip(): multiple lists ko pair banata hai
for name, score in zip(names, scores):
    print(f"{name} scored {score}")

# range(start, stop, step)
even_nums = list(range(2, 11, 2))  # [2, 4, 6, 8, 10]


# ==========================================
# 7. FUNCTIONAL PROGRAMMING: map(), filter(), lambda
# ==========================================
nums = [1, 2, 3, 4, 5, 6]

# map(): har element par operation lagata hai
squared = list(map(lambda x: x**2, nums))

# filter(): condition match hone par hi value rakhta hai
evens = list(filter(lambda x: x % 2 == 0, nums))

# all() & any(): check conditions across iterable
all_positive = all(x > 0 for x in nums)   # True (agar sab > 0 ho)
any_gt_five = any(x > 5 for x in nums)    # True (agar ek bhi > 5 ho)

print("Squared:", squared)
print("Evens:", evens)
print("All positive?:", all_positive)


# ==========================================
# 8. CUSTOM FUNCTIONS, *args, **kwargs
# ==========================================
def dynamic_calculator(operation, *args, **kwargs):
    """
    *args -> Tuple of positional arguments
    **kwargs -> Dictionary of keyword arguments
    """
    prefix = kwargs.get("prefix", "Result:")
    
    if operation == "multiply":
        res = 1
        for val in args:
            res *= val
        return f"{prefix} {res}"
    return f"{prefix} Unsupported operation"

print(dynamic_calculator("multiply", 2, 3, 4, prefix="Total Product:"))


# ==========================================
# 9. ERROR HANDLING (try, except, else, finally)
# ==========================================
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as err:
        return f"Error: Zero se divide nahi kar sakte ({err})"
    except TypeError as err:
        return f"Error: Data type invalid hai ({err})"
    else:
        # Tab chalta hai jab koi error na aaye
        return f"Success! Result = {result}"
    finally:
        # Hamesha run hoga
        pass

print(safe_divide(10, 2))
print(safe_divide(10, 0))


# ==========================================
# 10. FILE HANDLING (Context Manager: with)
# ==========================================
file_path = "sample_test.txt"

# Write mode ('w')
with open(file_path, "w", encoding="utf-8") as f:
    f.write("Line 1: Learning Python Built-ins\n")
    f.write("Line 2: Ready for Data Pipelines\n")

# Read mode ('r')
with open(file_path, "r", encoding="utf-8") as f:
    content = f.read()

print("File Read Output:\n" + content.strip())
```
